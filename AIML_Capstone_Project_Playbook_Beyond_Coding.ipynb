# AIML Capstone Project Playbook
## Beyond Coding — From Problem Statement to Defensible Solution

> **A capstone is not a collection of code. It is a structured process of understanding a problem, working with evidence, making decisions, building a solution, and communicating what the result means.**

This playbook is designed to help learners approach AIML capstone projects systematically.

It is **not a ready-made solution notebook**. It is a project-thinking and execution guide.

### The core mindset

```text
Business Problem
      ↓
Problem Definition
      ↓
Data Understanding
      ↓
EDA & Evidence
      ↓
ML Problem Formulation
      ↓
Approach & Model Selection
      ↓
Experimentation
      ↓
Evaluation
      ↓
Error Analysis
      ↓
Business Interpretation
      ↓
Recommendation / Action
      ↓
Documentation & Defense
```

### Projects used as examples

- **Capstone 1 — Autonomous Driving:** vehicle object detection + autopilot/road-safety analysis.
- **Capstone 2 — Preserving Heritage:** structure image classification + tourism analysis + recommendation engine.

The same thinking can be reused for other AIML projects.

# 1. Understand the Problem Before Writing Code

The first question is not:

> **"What code should I write?"**

It is:

> **"What problem am I trying to solve?"**

Convert the problem statement into:

```text
Business Context
      ↓
Business Problem
      ↓
Business Objective
      ↓
ML Objective
      ↓
Input
      ↓
Expected Output
      ↓
Success Criteria
```

### Project definition template

| Question | Your Answer |
|---|---|
| What is the business problem? | |
| What is the business objective? | |
| What is the ML objective? | |
| What is the input? | |
| What should the system output? | |
| How will success be measured? | |
| What is the expected business value? | |

### Examples

**Autonomous Driving — Part 1**

> Identify vehicle types and localize vehicles in road images.

This leads to **object detection** because the output requires both the vehicle class and its location.

**Heritage — Part 1**

> Predict the category of a historical structure from an image.

This leads to **image classification** because the required output is a category.

**Tourism — Part 2**

> Recommend places to tourists using user, place, and rating information.

This leads to a **recommendation-system** problem; the project specifies collaborative filtering.

### 60-second explanation

A learner should be able to say:

> "We are solving ________. The input is ________. The expected output is ________. We formulate it as a ________ problem and will measure success using ________."

If this cannot be explained clearly, coding should not start yet.

# 2. Understand the Dataset Before Modeling

Never begin modeling with a dataset you do not understand.

Ask:

### For tabular data

- What does one row represent?
- What does each column mean?
- Which columns are numeric, categorical, or dates?
- What is the target?
- How many rows and columns?
- Are there missing values?
- Are there duplicates?
- Are there anomalies?
- Are some columns irrelevant?
- Are there relationships between datasets?

### For image data

- How many images are available?
- What are the classes?
- How many images per class?
- Are images valid and readable?
- Are dimensions/formats consistent?
- Are labels correct?
- Are annotations available when localization is required?
- Are some classes severely under-represented?

### Capstone-specific thinking

**Capstone 1**

The vehicle-image component requires understanding the images **and the annotation information needed for object localization**. The `Tesla-Deaths.csv` component requires understanding what one row represents and what each safety-related field means.

**Capstone 2**

The image component requires understanding the training/test structure and classes. The tourism component uses:

```text
user.csv
   ↓ user_id
tourism_rating.csv
   ↓ place_id
tourisim_with_id.csv
```

Before joining datasets, identify the keys and understand whether they are unique.

# 3. Data Quality — Clean With a Reason

Cleaning is not about applying every available preprocessing technique.

Use:

```text
Inspect
  ↓
Identify Problem
  ↓
Investigate
  ↓
Decide
  ↓
Transform
  ↓
Validate
```

### Check for

- Missing values
- Duplicates
- Invalid values
- Anomalies
- Incorrect labels
- Poor-quality images
- Incorrect detection annotations
- Class imbalance
- Data leakage

### Ask before removing anything

> Why is this value wrong?

> Is it actually an error?

> What happens if I remove it?

> Could removing it introduce bias?

### Image/detection-specific checks

For object detection, verify:

- annotation exists
- class is correct
- bounding box is valid
- bounding box lies within the image
- the box actually covers the intended object

### Data leakage

Leakage occurs when information from outside the training process improperly influences the model.

Examples include:

- duplicate/near-duplicate images across train and test
- using test information during model development
- preprocessing using information from the complete dataset before splitting
- features that directly reveal the target

### Cleaning decision log

| Problem | Investigation | Decision | Reason | Impact |
|---|---|---|---|---|
| | | | | |
| | | | | |

> **Do not remove data just because a cleaning technique exists. Understand first, then decide.**

# 4. EDA — Ask Questions, Don't Make Random Charts

Exploratory Data Analysis is not a collection of graphs.

The goal is to find evidence that helps answer meaningful questions.

Use:

```text
Business Question
      ↓
Data Question
      ↓
Analysis
      ↓
Visualization
      ↓
Observation
      ↓
Insight
      ↓
Business Implication
```

### Useful EDA questions

**Capstone 1 — Safety**

- How do events vary over time?
- How are events distributed geographically?
- What patterns appear in victims/deaths?
- How often are cyclists/pedestrians involved?
- How are events distributed across Tesla models?
- What does the data show about verified Autopilot-related deaths?

**Capstone 2 — Tourism**

- What is the age distribution of users?
- Where do tourists originate?
- Which categories and locations are most represented?
- Which city is attractive for a nature enthusiast?
- Which places/cities/categories receive the strongest ratings?

### Observation vs Insight

**Observation**

> "Category A has a higher average rating than Category B."

**Insight**

> "Category A appears more positively rated in this dataset."

**Business implication**

> "The recommendation system could consider category preference when ranking places."

Do not jump directly from a chart to a causal claim.

### EDA checklist

```text
☐ Every chart answers a question
☐ Important distributions examined
☐ Important groups compared
☐ Trends investigated
☐ Relationships investigated
☐ Unexpected findings investigated
☐ Observations separated from interpretations
☐ Business implications documented
```

> **A good EDA tells a story about the data. It does not simply display the data.**

# 5. Formulate the ML Problem Correctly

The expected output determines the ML problem.

| Expected Output | Problem Type |
|---|---|
| One category | Classification |
| Class + object location | Object Detection |
| Numeric value | Regression |
| Personalized ranked items | Recommendation |
| Similar items/documents | Retrieval / Similarity |
| Groups without predefined labels | Clustering |

### The key questions

> **What exactly should the system predict, detect, rank, or retrieve?**

### Examples

```text
Structure image
      ↓
Structure category
      ↓
Image Classification
```

```text
Road image
      ↓
Vehicle class + bounding box
      ↓
Object Detection
```

```text
User + interactions
      ↓
Ranked places
      ↓
Recommendation
```

### Decision template

| Question | Answer |
|---|---|
| Input | |
| Expected output | |
| ML problem type | |
| Why this formulation? | |
| Candidate approaches | |
| Primary metric | |

### Important distinction

**Classification:** What is it?

**Object detection:** What is it and where is it?

**Recommendation:** What is likely to be relevant to this user?

> **Problem formulation comes before model selection.**

# 6. Choose the Approach — Justify, Don't Guess

Once the ML problem is clear, select an appropriate approach.

Use:

```text
Problem
  ↓
Data
  ↓
Constraints
  ↓
Candidate Approaches
  ↓
Baseline
  ↓
Experiment
  ↓
Final Choice
```

Consider:

- problem type
- dataset size and quality
- number of classes/items
- available compute
- training/inference requirements
- complexity
- expected performance
- deployment constraints
- business requirements

### Capstone examples

| Component | Problem | Reasonable approach |
|---|---|---|
| Heritage image category | Classification | CNN / pretrained CNN |
| Vehicle class + location | Object Detection | Detection architecture appropriate to dataset |
| Tourism recommendations | Recommendation | Collaborative filtering as specified by project |

### Avoid this

> "I selected this model because it is popular."

### Prefer this

> "We selected this approach because it is appropriate for the problem, dataset, and constraints. We will validate the choice experimentally."

### Model decision log

| Decision | Choice | Why? | Alternative | Evidence |
|---|---|---|---|---|
| Problem formulation | | | | |
| Baseline | | | | |
| Model/approach | | | | |
| Final selection | | | | |

> **The best model is not automatically the most complex model. It is the model that is appropriate and defensible for the problem.**

# 7. Model Development — Keep the Experiment Structured

Once the approach is selected, move from baseline to experimentation.

```text
Baseline
   ↓
Train
   ↓
Validate
   ↓
Observe
   ↓
Diagnose
   ↓
Change an important factor
   ↓
Train Again
   ↓
Compare
```

### For image classification

A practical path is:

```text
Images
  ↓
Class / quality checks
  ↓
Train-validation-test strategy
  ↓
Baseline
  ↓
CNN / Transfer Learning
  ↓
Optional augmentation
  ↓
Validation
  ↓
Error analysis
```

For the heritage project, transfer learning and augmentation are project-relevant experiments.

### For object detection

```text
Images + annotations
        ↓
Validate annotations
        ↓
Visualize boxes
        ↓
Train detector
        ↓
Validate
        ↓
Evaluate
        ↓
Visualize predictions
        ↓
Error analysis
```

### For recommendation

```text
Users + Places + Ratings
        ↓
Understand interactions
        ↓
Build baseline
        ↓
Collaborative filtering
        ↓
Generate ranked recommendations
        ↓
Evaluate
```

### Experiment log

| Experiment | Approach | Important Change | Validation Result | Observation |
|---|---|---|---|---|
| E1 | Baseline | | | |
| E2 | | | | |
| E3 | | | | |

> **Do not change five things at once. A useful experiment should answer a question.**

# 8. Evaluation — How Do You Know the Solution Works?

A model is not "good" simply because it produced a high number.

The evaluation must match the problem.

| Problem | Useful evaluation |
|---|---|
| Classification | Accuracy, Precision, Recall, F1, confusion matrix |
| Object Detection | IoU, Precision, Recall, mAP |
| Recommendation | Ranking/relevance measures such as Precision@K or Recall@K, plus practical usefulness |

### Ask three questions

**1. What are we measuring?**

**2. Why is this metric appropriate?**

**3. What does the result mean in the context of the problem?**

### Classification

Do not stop at overall accuracy.

Ask:

- Which classes perform well?
- Which classes are confused?
- Is there class imbalance?
- What does the confusion matrix show?

### Object Detection

Consider both:

```text
Correct object class
        +
Correct localization
```

IoU helps assess overlap between predicted and ground-truth bounding boxes.

### Recommendation

A recommendation system should ultimately produce useful rankings, not just predictions.

Ask:

- Are the recommendations relevant?
- Are they personalized?
- How will ranking quality be evaluated?
- Does the result make sense for the user scenario?

### Test-set discipline

Use validation data to guide development.

Reserve the test set for final evaluation.

> **A metric is meaningful only when you understand what it measures and why it matters.**

# 9. Error Analysis — Learn From What the Model Gets Wrong

The most valuable question after evaluation is often:

> **"Where does the system fail?"**

Use:

```text
Result
  ↓
Failure Cases
  ↓
Pattern
  ↓
Possible Cause
  ↓
Improvement
```

### Classification

Investigate:

- confused classes
- poor-quality images
- minority classes
- incorrect labels
- visually similar categories

### Object Detection

Investigate:

- false positives
- missed objects
- wrong classes
- poor bounding boxes
- small/distant objects
- crowded or partially visible objects

### Recommendation

Investigate:

- irrelevant recommendations
- insufficient interaction data
- cold-start situations
- overly popular recommendations
- lack of personalization

### Error analysis table

| Failure | Where observed? | Possible cause | What could improve it? |
|---|---|---|---|
| | | | |
| | | | |

> **A strong project does not hide failure cases. It explains them.**

A final presentation should include at least a few representative failure cases and what was learned from them.

# 10. From Results to Business Insight

This is where an AIML project moves **beyond coding**.

A model output is not automatically a business insight.

Use:

```text
Result
  ↓
Interpretation
  ↓
Insight
  ↓
Business Implication
  ↓
Action / Recommendation
```

### Example

**Result**

> A particular tourism category receives higher ratings in the dataset.

**Interpretation**

> Users appear to respond more positively to that category.

**Business implication**

> The category may deserve greater consideration in recommendation rankings.

**Action**

> Test whether including category preference improves recommendation relevance.

### Important distinction

```text
Data says:
____________________

Therefore we can reasonably infer:
____________________

Business implication:
____________________

Recommended action:
____________________
```

Do not make claims that the data cannot support.

For observational data especially:

> **Association does not automatically establish causation.**

For example, an increase in reported Tesla-related events does not by itself prove that Autopilot caused the increase or became less safe.

Consider factors such as exposure, vehicle population, reporting practices, and other contextual variables before making causal claims.

> **The purpose of ML is not only prediction. It is to support better decisions.**

# 11. Recommendation Systems — Think in Users, Items and Interactions

A recommendation system answers a different question from classification.

It asks:

> **"What should this user be shown, suggested, or offered?"**

For the tourism component:

```text
Users
   +
Places
   +
Ratings
   ↓
User–Item Interactions
   ↓
Collaborative Filtering
   ↓
Ranked Recommendations
```

### Think about the data

Before building the recommender, understand:

- users
- places/items
- ratings/interactions
- user-item relationships
- sparsity
- users/items with little or no interaction data

### Recommendation workflow

```text
Inspect datasets
      ↓
Validate keys
      ↓
Combine relevant data
      ↓
Understand rating behaviour
      ↓
Establish baseline
      ↓
Build collaborative filtering approach
      ↓
Generate recommendations
      ↓
Evaluate relevance/ranking
      ↓
Inspect examples
```

### Questions for the project

- What makes a recommendation useful?
- Is it personalized?
- How will recommendations be ranked?
- How will you evaluate them?
- What happens when a new user has no ratings?
- What happens when an item has very few ratings?

> **A recommendation engine is not complete just because it returns a list. The list must be meaningful for the user.**

A capstone should preserve the reasoning behind important decisions.

Create a **Decision Log** throughout the project.

| Data cleaning | | | | |
| EDA | | | | |
| Model selection | | | | |
| Training approach | | | | |
| Evaluation metric | | | | |

### Why this matters

During a viva, the question is often not:

> "Can you run the code?"

It is:

> **"Why did you do this?"**

The decision log prepares you to answer that question.

A strong project can explain:

- what was done
- why it was done
- what alternatives were considered
- what evidence supported the decision
- what happened after the decision

# 13. Tell the Story of the Project

A good capstone presentation should not be a sequence of code screenshots.

Tell a coherent story:

```text
1. Problem
      ↓
2. Why it matters
      ↓
3. Data
      ↓
4. What we discovered
      ↓
5. ML formulation
      ↓
6. Approach
      ↓
7. Experiments
      ↓
8. Evaluation
      ↓
9. Failure cases
      ↓
10. Business insight
      ↓
11. Recommendation
      ↓
12. Limitations
      ↓
13. Future improvement
```

### A useful project narrative

> **We started with...**

> **The data showed...**

> **Therefore, we formulated the problem as...**

> **We selected this approach because...**

> **Our experiments showed...**

> **The model performed well/poorly in...**

> **The important business insight is...**

> **The main limitation is...**

> **If we had more data/time, we would...**

This is much stronger than presenting only the final accuracy.

# 14. Project Defense — Questions You Should Be Ready For

Before submitting the project, make sure you can answer:

1. What problem are you solving?
2. Why is this problem important?
3. What does one data point/image represent?
4. What is your expected output?
5. Why did you formulate it as this ML problem?
6. What assumptions did you make?
7. What data-quality issues did you find?
8. How did you prevent data leakage?
9. Why did you select this approach/model?
10. What alternatives did you consider?
11. What evaluation metric did you choose and why?
12. Where does the solution fail?
13. What did the error analysis reveal?
14. What does the result mean for the business?
15. What are the major limitations?
16. What would you improve with more data or time?
17. How could this solution be deployed?
18. What would you monitor after deployment?

### The strongest answer pattern

```text
Decision
   ↓
Reason
   ↓
Evidence
   ↓
Result
   ↓
Limitation
```

Do not be afraid to say:

> **"We did not have enough evidence to conclude that."**

Knowing the limitation of your analysis is part of technical maturity.

# Final Question

Before calling the capstone complete, ask:

> **"Can I explain not only what I built, but why I built it, how I know it works, where it fails, and what someone should do with the result?"**

If the answer is **yes**, you are thinking beyond coding.


# The Capstone Mindset

A weak capstone asks:

> **"What code should I write?"**

A better capstone asks:

> **"What model should I use?"**

A strong capstone asks:

> **"What problem am I solving, what evidence do I have, why did I choose this approach, how do I know it works, where does it fail, and what should someone do with the result?"**
